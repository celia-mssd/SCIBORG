```python
import pandas as pd
import numpy as np

def load_interactions(file_path):
    "Loads interactions from networks.csv file"
    "Returns list of interactions"
    df = pd.read_csv(file_path)
    interactions = []

    for _, row in df.iterrows():  
        network = {rule for rule, presence in row.items() if presence == 1} # Garder seulement les règles activées
        interactions.append(network)  # Ajouter le réseau traité
    
    return interactions

def jaccard_index(networks_1, networks_2, weighted=False):
    "Computes Jaccard index between two ensembles of boolean rules"
    from collections import Counter
    
    if weighted:

        set_1 = Counter([rule for network in networks_1 for rule in network])
        set_2 = Counter([rule for network in networks_2 for rule in network])
        all_rules = set(set_1) | set(set_2)
        
        intersection = sum(min(set_1[r], set_2[r]) for r in all_rules)
        union = sum(max(set_1[r], set_2[r]) for r in all_rules)
        if union==0:
            print('union = 0')
            index = 0
        else :
            index=intersection/union
        return index, intersection,union,all_rules
    
    else:
        set_1 = set().union(*networks_1)
        set_2 = set().union(*networks_2)

        intersection = [set_1 & set_2]
        union = [set_1 | set_2]
        len_intersection = len(set_1 & set_2)
        len_union = len(set_1 | set_2)
        index = len_intersection / len_union if len_union != 0 else 0
        return index, intersection, union
```
